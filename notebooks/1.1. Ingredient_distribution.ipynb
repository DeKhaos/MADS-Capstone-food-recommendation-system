```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df=pd.read_csv("ingredient_counts_ingredients_canonical_final_le_5_replace.csv")
```

```python
df.columns
```

```python
df_head = df.head(10).copy()
df_tail = df.tail(10).copy()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 6))


ax1.bar(df_head['Ingredient'], df_head['Count'], color='skyblue')
ax1.set_title('Top 10 Ingredients')
ax1.set_ylabel('Count')
ax1.tick_params(axis='x', rotation=45)


ax2.bar(df_tail['Ingredient'], df_tail['Count'], color='salmon')
ax2.set_title('Bottom 10 Ingredients')
ax2.set_ylabel('Count')
ax2.tick_params(axis='x', rotation=82)
plt.savefig('ingredient_chart.png',dpi=300, bbox_inches='tight')
```
